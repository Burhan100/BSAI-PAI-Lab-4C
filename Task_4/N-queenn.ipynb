```python
# Requirements:
# Python 3.x

# Description:
# Console-based visualization of the N-Queen problem.
# The user manually places queens and the program checks
# whether the position is safe or not.

# Purpose:
# To understand the N-Queen problem using logical conditions,
# loops, and board visualization.


# Function to display the chessboard
def display_board(board, n):
    print("\nChessboard:")
    for r in range(n):
        for c in range(n):
            print("|", board[r*n + c], end=" ")
        print("|")
    print()


# Function to check if the position is safe
def is_safe(board, n, row, col):

    # Check same row
    for c in range(n):
        if board[row*n + c] == 'Q':
            return False

    # Check same column
    for r in range(n):
        if board[r*n + col] == 'Q':
            return False

    # Check upper-left diagonal
    r = row
    c = col
    while r >= 0 and c >= 0:
        if board[r*n + c] == 'Q':
            return False
        r -= 1
        c -= 1

    # Check upper-right diagonal
    r = row
    c = col
    while r >= 0 and c < n:
        if board[r*n + c] == 'Q':
            return False
        r -= 1
        c += 1

    # Check lower-left diagonal
    r = row
    c = col
    while r < n and c >= 0:
        if board[r*n + c] == 'Q':
            return False
        r += 1
        c -= 1

    # Check lower-right diagonal
    r = row
    c = col
    while r < n and c < n:
        if board[r*n + c] == 'Q':
            return False
        r += 1
        c += 1

    return True


# Main program
while True:

    n = int(input("Enter board size (4, 5, or 6): "))

    if n not in [4, 5, 6]:
        print("Invalid board size. Try again.")
        continue

    # Create empty board
    board = [' '] * (n*n)

    queens = 0

    display_board(board, n)

    # Place queens
    while queens < n:

        pos = int(input(f"Enter position to place queen (1-{n*n}): "))

        if pos < 1 or pos > n*n:
            print("Invalid position.")
            continue

        pos = pos - 1
        row = pos // n
        col = pos % n

        if board[pos] == 'Q':
            print("Position already occupied!")
            continue

        if is_safe(board, n, row, col):
            board[pos] = 'Q'
            queens += 1
            display_board(board, n)
        else:
            print("Invalid move! Queens attack each other.")
            break

    if queens == n:
        print("Success! All queens placed safely.")

    restart = input("Restart game? (y/n): ")

    if restart.lower() != 'y':
        print("Game Ended.")
        break
```
